
# To perform ML by using  dimensionality reduction techniques PCA , TSNE (Unsuperwised learning)



```python
import pandas as pd
```

```python
df = pd.read_csv('/content/data.csv')
```

```python
del df['PassengerId']
del df['Ticket']
del df['Fare']
del df['Cabin']
del df['Name']
df['Age'] = df['Age'].fillna(df['Age'].mean())

df = df.dropna()
df['Age'] = df['Age'].astype(int)
df = pd.get_dummies(df,columns= ['Sex','Embarked','Pclass'],drop_first=False,dtype=int)
df['No.of_peoples'] = df['Parch'] + df['SibSp']
del df['SibSp']
del df['Parch']
df['Age'] = df['Age']/df['Age'].max()
df['No.of_peoples'] = df['No.of_peoples']/df['No.of_peoples'].max()
y = df['Survived']
del df['Survived']
x = df
```

```python
x
```

```python
from sklearn.decomposition import PCA

n = 3
pca = PCA(n_components=n)
columns = ['pca_'+str(i+1) for i in range(n) ]
x_pca = pca.fit_transform(x)

x_pca_df = pd.DataFrame(x_pca , columns = columns)

x_pca_df
```

```python
x_train = x_pca_df[:650]
x_test  = x_pca_df[650:]

y_train = y[:650]
y_test  = y[650:]

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(x_train,y_train)
print("Model trained successfully for PCA")
```

```python
accuracy = model.score(x_train,y_train)
print(f"Training Accuracy of the Model: {round( accuracy*100 , 2 )} % " )



# Testing Accuracy
accuracy = model.score(x_test,y_test)
print(f"Testing Accuracy of the Model  : {round( accuracy*100 , 2 )} % " )
```

By using TSNE

```python
from sklearn.manifold import TSNE
n = 3
tsne = TSNE(n_components=n)
columns = ['tsne_'+str(i+1) for i in range(n) ]
x_tsne = tsne.fit_transform(x)

x_tsne_df = pd.DataFrame(x_tsne , columns = columns)

x_pca_df
```

```python
x_train = x_tsne_df[:650]
x_test  = x_tsne_df[650:]

y_train = y[:650]
y_test  = y[650:]

from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)

model.fit(x_train,y_train)
print("Model trained successfully for TSNE")
```

```python
accuracy = model.score(x_train,y_train)
print(f"Training Accuracy of the Model: {round( accuracy*100 , 2 )} % " )



# Testing Accuracy
accuracy = model.score(x_test,y_test)
print(f"Testing Accuracy of the Model  : {round( accuracy*100 , 2 )} % " )
```

-- Whats PCA and why is it       supervised learning

PCA (Principal Component Analysis) is one of the most commonly used **techniques** for dimensionality reduction in Machine Learning .
PCA takes your original features and creates new features called Principal Components.

PCA doesn't need a target/label y instead it focuses on representing the features using fewer dimensions while retaining as much variance(variation) as possible
